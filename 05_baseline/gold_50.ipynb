```python
import pandas as pd
import soundfile as sf
from pathlib import Path
from IPython.display import Audio, display

LOYIHA = Path("..")   # repo ildizi (notebook 05_baseline papkasidan ishga tushiriladi)

jadval = pd.read_csv(LOYIHA / "05_baseline" / "eval_50.csv")
natija_fayli = LOYIHA / "05_baseline" / "gold_50.csv"

# oldin belgilaganlaring bo'lsa, ular yuklanadi
if natija_fayli.exists():
    belgilar = pd.read_csv(natija_fayli).set_index("idx").to_dict("index")
else:
    belgilar = {}

print(len(jadval), "ta klip,", len(belgilar), "tasi tekshirilgan")
```

```python
def keyingi():
    # hali tekshirilmagan birinchi klipning tartib raqamini topadi
    for raqam in range(len(jadval)):
        if int(jadval.iloc[raqam]["idx"]) not in belgilar:
            return raqam
    print("Hammasi tekshirildi!")
    return None

def korsat(raqam):
    qator = jadval.iloc[raqam]
    print(f"[{raqam + 1}/50]  idx = {qator['idx']}")
    print("Asl matn:   ", qator["matn"])
    print("Model uchun:", qator["matn_model"])
    audio_yoli = LOYIHA / str(qator["fayl"]).replace("\\", "/")
    ovoz, sr = sf.read(str(audio_yoli))   # faylni o'qiymiz
    if ovoz.ndim > 1:                     # stereo bo'lsa, shaklini to'g'rilaymiz
        ovoz = ovoz.T
    display(Audio(ovoz, rate=sr))

def belgila(raqam, holat, togri_matn="", izoh=""):
    # holat: "togri", "xato" yoki "audio"
    if holat not in ["togri", "xato", "audio"]:
        print("Holat faqat: togri, xato yoki audio")
        return
    idx = int(jadval.iloc[raqam]["idx"])
    belgilar[idx] = {"holat": holat, "togri_matn": togri_matn, "izoh": izoh}
    pd.DataFrame.from_dict(belgilar, orient="index").rename_axis("idx").to_csv(natija_fayli)
    print("Saqlandi:", len(belgilar), "/ 50")
```

```python
r = keyingi()
if r is not None:   # hammasi tekshirilgan bo'lsa, ko'rsatadigan klip yo'q
    korsat(r)
```
